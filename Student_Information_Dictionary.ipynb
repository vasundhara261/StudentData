```python
#Studnet_Information_Dictionary
students = {
  101:{
    "name":"Rahul",
    "age":20,
    "course":"Data Science",
    "marks":85
},
  102:{
      "name":"Priya",
      "age":21,
      "course":"cse",
      "marks":90,
  },
  103:{
      "name":"Arun",
      "age":20,
      "course":"AI",
      "marks":95,
  }
}
print(students)

```

```python
def search_students(students_id, students_dict):
    if students_id in students_dict:
        print("students Found!")
        print("ID:", students_id)
        print("Name:", students_dict[students_id]["name"])
        print("Age:", students_dict[students_id]["age"])
        print("Course:", students_dict[students_id]["course"])
        print("Marks:", students_dict[students_id]["marks"])
    else:
        print("Students not found!")
```

```python
search_students(102, Students)
```

```python
def add_student(student_id, name, age, course, marks, students_dict):
    if student_id in students_dict:
        print("Student ID already exists!")
    else:
        students_dict[student_id] = {
            "name": name,
            "age": age,
            "course": course,
            "marks": marks
        }
        print("Student added successfully!")
```

```python
add_student(104, "Sneha", 20, "Data Science", 92, Students)
```

```python
search_students(104, Students)    #To verify that the student was added
```

```python
def update_student(student_id, field, new_value, students_dict):
    if student_id in students_dict:
        if field in students_dict[student_id]:
            students_dict[student_id][field] = new_value
            print("Student information updated successfully!")
        else:
            print("Invalid field!")
    else:
        print("Student not found!")
```

```python
update_student(101, "marks", 95, Students)
```

```python
def display_students(students_dict):
    if not students_dict:
        print("No student records available.")
    else:
        for student_id, details in students_dict.items():
            print("\nStudent ID:", student_id)
            print("Name:", details["name"])
            print("Age:", details["age"])
            print("Course:", details["course"])
            print("Marks:", details["marks"])
            print("-" * 30)
```

```python
display_students(Students)
```
